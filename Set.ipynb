```python
my_set = {10, 20, 30}

my_set.add(40)

print(my_set)
```

```python
my_set = {10, 20, 30}

my_set.discard(20)

print(my_set)
```

```python
my_set = {10, 20, 30, 40}

if 20 in my_set:
    print("20 is present")
else:
    print("20 is not present")
```

```python
set1 = {1, 2, 3}
set2 = {3, 4, 5}

new_set = set1.union(set2)

print(new_set)
```

```python
set1 = {1, 2, 3}
set2 = {2, 3, 4}

new_set = set1.intersection(set2)

print(new_set)
```

```python
set1 = {1, 2, 3}
set2 = {2, 3, 4}

new_set = set1.difference(set2)

print(new_set)
```
